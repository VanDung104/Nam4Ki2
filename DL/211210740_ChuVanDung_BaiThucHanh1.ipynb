```python
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import BinaryCrossentropy


```

```python
from keras.datasets.mnist import load_data


(X_train, y_train), (X_test, y_test) = load_data()
```

```python
X_train.shape
```

```python
X_train_reshaped = X_train.reshape(-1, 784)
```

```python
from tensorflow.keras.utils import to_categorical

y_train = to_categorical(y_train, num_classes = 10)
```

```python
from keras.models import Sequential


model = Sequential()
```

```python
Dense(784, input_shape=(784, ))
```

```python
model.add(Dense(784, input_shape=(784, )))
```

```python
model.add(Dense(10, input_shape=(10, )))
```

```python
model.summary()
```

```python
optimizer = Adam(learning_rate=0.01)
```

```python
loss = BinaryCrossentropy()
```

```python
model.compile(optimizer=optimizer,loss=loss, metrics=['accuracy'])
```

```python
model.fit(X_train_reshaped, y_train, batch_size=128, epochs=10)
```

```python
X_test_reshaped = X_test.reshape(-1, 784)
```

```python
y_pred = model.predict(X_test_reshaped)
```

```python
y_pred.shape
```

```python
import numpy as np
```

```python
y_pred = np.argmax(y_pred, axis = -1)
```

```python
from sklearn.metrics import accuracy_score


accuracy_score(y_test, y_pred)*100
```

#### xây model

```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, MaxPool2D, Flatten
```

### 2 lớp cơ bản

```python

model5_1 = Sequential([
    # Lớp 1: Conv -> Pooling
    Conv2D(filters=32, kernel_size=(4,4), activation='relu', input_shape=(32,32,3)),
    MaxPool2D(pool_size=(2,2)),

    # Lớp 2: Fully Connected (Output)
    Flatten(),
    Dense(10, activation='softmax')  # 10 lớp đầu ra
])

model5_1.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

model5_1.summary()
```

```python
history = model5_1.fit(X_train_5, y_train_cat_5,
                    epochs=10,
                    validation_data=(X_test_5, y_test_cat_5))
```

```python
# Đánh giá độ chính xác
test_loss, test_acc = model5_1.evaluate(X_test_5, y_test_cat_5, verbose=0)
print(f'Độ chính xác trên tập Test: {test_acc*100:.2f}%')
print(f'Loss test: {test_loss:.4f}')
```

```python
model = Sequential()
model.add(Conv2D(filters=32, kernel_size=(4,4), input_shape=(32,32,3),activation='relu'))
model.add(MaxPool2D(pool_size=(2,2)))

model.add(Conv2D(filters=32, kernel_size=(4,4), input_shape=(32,32,3),activation='relu'))
model.add(MaxPool2D(pool_size=(2,2)))

model.add(Flatten())
model.add(Dense(256,activation='relu'))
model.add(Dense(10,activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam',
             metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(X_train_5, y_train_cat_5,
                    epochs=10,
                    validation_data=(X_test_5, y_test_cat_5))
```

```python
model.evaluate(X_test_5,y_test_cat_5,verbose=0)
```
