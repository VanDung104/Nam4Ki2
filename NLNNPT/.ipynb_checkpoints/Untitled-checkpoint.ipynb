```python
import math

def next_term(x, n, term):
    return term * (-x**2 / ((2*n) * (2*n + 1)))

def sin_taylor(x, epsilon=0.01):
    def iter_terms():
        """Generator sinh từng hạng tử trong chuỗi Taylor"""
        n = 1
        term = x  # Hạng tử đầu tiên là x
        while True:
            yield term
            term = next_term(x, n, term)  # Tính hạng tử tiếp theo
            n += 1

    def within(epsilon, iterable):
        """Tìm giá trị hội tụ của chuỗi khi sai số giữa 2 lần lặp < epsilon"""
        total = 0 
        prev = 0 
        for term in iterable:
            total += term
            if abs(total - prev) <= epsilon:
                return total
            prev = total  # Cập nhật giá trị trước đó

    return within(epsilon, iter_terms())  # Bắt đầu tính chuỗi


x = math.radians(30)
epsilon = 0.0001
approx_value = sin_taylor(x, epsilon)
true_value = math.sin(x)
print(f"Giá trị gần đúng của sin(30°): {approx_value}")
print(f"Giá trị thực của sin(30°): {true_value}")
```

```python
def phi_euler(n):
    result = 1  # Khởi tạo giá trị kết quả
    p = 2  # Bắt đầu từ số nguyên tố đầu tiên

    while p * p <= n:  # Lặp qua tất cả các số nguyên tố ≤ sqrt(n)
        if n % p == 0:
            k = 0
            while n % p == 0:
                n //= p
                k += 1  # Đếm số mũ k của p trong n
            result *= (p ** k - p ** (k - 1))  # Áp dụng công thức p^k - p^(k-1)
        p += 1

    if n > 1:  # Nếu còn lại một số nguyên tố lớn hơn 1
        result *= (n - 1) 

    return result

# Kiểm tra
print(phi_euler(10)) 
print(phi_euler(100))
print(phi_euler(81))
print(phi_euler(13))
```

```python
def read_points(filename):
    with open(filename) as f:
        for line in f:
            yield tuple(map(float, line.split()))
```

```python
points = read_points("map.txt")
```

```python
def triplets(iterable):
    def triplet_from(first, second, iterable_tail):
        try:
            third = next(iterable_tail)
            yield first, second, third
            yield from triplet_from(second, third, iterable_tail)
        except StopIteration:
            return  

    try:
        iterator = iter(iterable)
        return triplet_from(next(iterator), next(iterator), iterator)  #
    except StopIteration:
        return  
```

```python
trips_iter = triplets(read_points("map.txt"))
```

```python
for trip in trips_iter:
    print(trip)
```

```python
import math

def chuvi(point1, point2, point3):
    def distance(p1, p2):
        return math.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)
    
    return distance(point1, point2) + distance(point2, point3) + distance(point3, point1)

```

```python
trips_iter = triplets(read_points("map.txt"))
quad =((triplet[0], triplet[1], triplet[2],chuvi(*triplet)) for triplet in trips_iter)
```

```python
for quad_iter in quad:
    print(quad_iter)
```

```python
from collections import namedtuple
Trip = namedtuple("Trip", ("x", "y","z"))
def series(n, row_iter):
    for row in row_iter:
        yield Trip(*row[n*3:n*3+3])
```

```python
with open("Anscombe.csv") as source:
    data = tuple(head_split_fixed(row_iter(source)))
    sample_I= tuple(series(0,data))
    sample_II= tuple(series(1,data))
    sample_III= tuple(series(2,data))
    sample_IV= tuple(series(3,data))
```

```python
mean = sum(float(trip.z) for trip in sample_I)/len(sample_I)
```

```python
 def s0(data):
    return sum(1 for x in data) # or len(data)
 def s1(data): 
    return sum(x for x in data) # or sum(data)
 def s2(data): 
    return sum(x*x for x in data)
```

```python
def mean1(x): 
    return s1(x)/s0(x)
def stdev1(x):
    return math.sqrt(s2(x)/s0(x) - (s1(x)/s0(x))**2)
```

```python
 def standard( x, μ_x, σ_x ):
    return (x-μ_x)/σ_x
```

```python
import math
    
```

```python
i = 1
for quy in sample_I,sample_II,sample_III,sample_IV:
    z_values = [float(trip.z) for trip in quy]
    mean_z, std_z = mean1(z_values), stdev1(z_values)
    z_standard = [round(standard(z,mean_z,std_z),3) for z in z_values]
    print(f"Mean(z) quý {i}: {mean_z:.3f}")
    print(f"Stdev(z) của quý {i}: {std_z:.3f}")
    print(f"danh sách các giá trị z sau khi chuẩn hóa quý {i}:", z_standard)
    print("-----------------------------------------------------------------")
    i +=1
```
